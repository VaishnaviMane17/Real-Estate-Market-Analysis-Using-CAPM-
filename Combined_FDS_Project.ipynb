````python
def daily_return(df):

  returns = df.copy()

  # Loop through each stock
  for i in df.columns[0:]:

    # Loop through each row belonging to the stock
    for j in range(1, len(df)):

      # Calculate the percentage of change from the previous day
      returns[i][j] = ((df[i][j]- df[i][j-1])/df[i][j-1]) * 100

    # set the value of first row to zero, as previous value is not available
    returns[i][0] = 0
  return returns
````

# Importing Risk Free Rate from  Fred API

The risk-free rate of return is the theoretical rate of return of an investment with zero risk. In real estate, it is often considered to be the yield on a long-term US Treasury bond. This is because US Treasury bonds are considered to be one of the safest investments in the world, and are often used as a benchmark for measuring the risk of other investments. The risk-free rate is used in finance to calculate the expected return of an investment after adjusting for the level of risk
